```python
import pandas as pd

# Sample dataset similar to tweet engagement analysis from the paper
data = {
    'tweet_id': [1, 2, 3, 4],
    'article_title': [
        "Digital Scholarship Trends",
        "Social Media in Academia",
        "Altmetric Analysis Study",
        "Networked Scholarship"
    ],
    'retweets': [15, 40, 22, 9],
    'likes': [100, 250, 180, 75],
    'mentions': [5, 12, 7, 3]
}

df = pd.DataFrame(data)

# Calculate a simple "engagement score"
df['engagement_score'] = df['retweets'] * 2 + df['likes'] + df['mentions'] * 3

print("Scholarly Communication Engagement Scores:\n")
print(df[['article_title', 'engagement_score']])

# Identify most influential article
top_article = df.loc[df['engagement_score'].idxmax()]
print("\nMost Influential Article:")
print(top_article['article_title'])
```
